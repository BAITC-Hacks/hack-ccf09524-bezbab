```python
from pathlib import Path
import pandas as pd

# Ищем data в текущей папке или среди её родительских папок.
data_dir = next(
    (parent / 'data' for parent in (Path.cwd(), *Path.cwd().parents) if (parent / 'data').is_dir()),
    None,
)
if data_dir is None:
    raise FileNotFoundError(f'Папка data не найдена от {Path.cwd()} и выше')

csv_files = sorted(data_dir.glob('*.csv'))
if not csv_files:
    raise FileNotFoundError(f'В папке {data_dir} нет CSV-файлов')

# Ключ словаря — имя файла без расширения, значение — DataFrame.
data = {file.stem: pd.read_csv(file, encoding='utf-8-sig') for file in csv_files}

print(f'Загружено {len(data)} таблиц из {data_dir.resolve()}')
pd.DataFrame([{'table': name, 'rows': len(df), 'columns': len(df.columns)} for name, df in data.items()])
```
